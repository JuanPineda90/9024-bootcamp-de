```sql
select count(*) from samples.nyctaxi.trips;
```

```sql
DESCRIBE samples.nyctaxi.trips;
```

```sql
select * 
from samples.nyctaxi.trips
where trip_distance between 0 and 2
order by trip_distance asc;
```

```sql
select * 
from samples.nyctaxi.trips
where trip_distance = 1
or trip_distance = 2;
```

```sql
select * 
from samples.nyctaxi.trips
where trip_distance in ('2','3');
```

```sql
select * 
from samples.nyctaxi.trips
where trip_distance = 3
and pickup_zip = 10110;
```

```sql
select pickup_zip, round(avg(trip_distance),2) as avg_trip
from samples.nyctaxi.trips
group by pickup_zip
order by pickup_zip;
```

```sql
select pickup_zip, count(*) as trip_count
from samples.nyctaxi.trips
group by pickup_zip
having trip_count > 1000
order by pickup_zip;
```
